# RMSD versus native-contact Jaccard

This notebook reads already-computed per-design metrics. AF3 supplies only Top10 membership; plotted RMSD and Jaccard values come from matching OpenMM-repacked PDBs.

```python
from pathlib import Path
import os, sys

ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
CONFIG_PATH = Path(os.environ.get('STRUCTURAL_METRICS_CONFIG', ROOT / 'config.yaml'))
print(f'Repository: {ROOT}')
print(f'Configuration: {CONFIG_PATH}')
```

```python
from IPython.display import display, Markdown
import matplotlib.pyplot as plt
from src.configuration import load_config
from src.plot_results import create_method_figures

config = load_config(CONFIG_PATH)
figures, top10_qc, unmatched, population = create_method_figures(config)
display(Markdown(f'**QC-passing background designs:** {len(population):,}  '))
display(top10_qc)
if len(unmatched):
    display(Markdown('**Unmatched AF3 Top10 identities**'))
    display(unmatched)
for method, figure in figures.items():
    display(Markdown(f'## {method}'))
    display(figure)
    plt.close(figure)
```

## Interpretation

Lower RMSD means more native-like peptide placement and shape in the receptor frame. Higher Jaccard means closer reproduction of the native residue-contact set. Stars ask whether AF3 ranking enriches for designs that were already more native-like in their OpenMM-repacked structures; they do not measure AF3 remodeling.
